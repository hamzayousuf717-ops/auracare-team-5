```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
from google.colab import files
uploaded = files.upload()
warnings.filterwarnings('ignore')
df=pd.read_csv('insurance.csv')
#EDA
df.shape
df.head()
df.info()
df.describe()
df.isnull().sum()
df.columns
numeric_colums=['age','bmi','children','charges']
for col in numeric_colums:
  plt.figure(figsize=(6,4))
  sns.histplot(df[col],kde=True,bins=20)
  sns.countplot(x=df['children'])
  sns.countplot(x=df['sex'])
for col in  numeric_colums:
  plt.figure(figsize=(6,4))
  sns.boxplot(x=df[col])
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),annot=True)
df_cleaned=df.copy()
#Checking duplicates
df_cleaned.shape
df_cleaned.drop_duplicates(inplace=True)
df_cleaned.shape
# checking is there are any null values in the data
df_cleaned.isnull().sum()
# for checking data types
df_cleaned.dtypes
# checking the value counts to see how many types of categorical data is present
df_cleaned['sex'].value_counts()
#label encoding
df_cleaned['sex']=df_cleaned['sex'].map({"male":0,"female":1})
df_cleaned.head()
df_cleaned['smoker'].value_counts()
df_cleaned['smoker']=df_cleaned['smoker'].map({"yes":0,"no":1})
# renaming
df_cleaned.rename(columns={'sex':'is_female',
                           'smoker':'is_smoker'
                           },inplace=True)
df_cleaned.head()
df_cleaned['region'].value_counts()
# one hot encoding
df_cleaned= pd.get_dummies(df_cleaned,columns=['region'],drop_first=True)
df_cleaned.head()
# this is so we can get 0 and 1
df_cleaned=df_cleaned.astype(int)
df_cleaned.head()
#Feature engineering and extraction
sns.histplot(df['bmi'])
#creating bins and new rows
df_cleaned['bmi_category']=pd.cut(df_cleaned['bmi'],bins=[0,18.5,24.9,29.9,float('inf')],
                                  labels=['underweight','normal','overweight','obese']
                                  )
df_cleaned
#one hot encoading
df_cleaned= pd.get_dummies(df_cleaned,columns=['bmi_category'],drop_first=True)
df_cleaned.head()
df_cleaned=df_cleaned.astype(int)
df_cleaned.head()
# for scaling
from sklearn.preprocessing import StandardScaler
cols=['age','bmi','children']
scaler=StandardScaler()
df_cleaned[cols]=scaler.fit_transform(df_cleaned[cols])
df_cleaned.head()
#pearson correlation
from scipy.stats import pearsonr
selected_features=[
    'age','bmi','is_female','is_smoker','children','region_northwest','region_southeast','region_southwest','bmi_category_normal','bmi_category_overweight','bmi_category_obese'

]
correlations={
    feature: pearsonr(df_cleaned[feature],df_cleaned ['charges'])[0]
    for feature in selected_features
}
correlation_df=pd.DataFrame(list(correlations.items()),columns=['Feature','pearson correlation'])
correlation_df.sort_values(by='pearson correlation',ascending=False)
#chi2 algo correlation
cat_features=[
    'bmi_category_obese','region_southeast','region_northwest','region_southwest','is_female','bmi_category_normal','bmi_category_overweight',	'is_smoker'
]
from scipy.stats import chi2_contingency
import pandas as pd
alpha=0.05
df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results={}
for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }
chi2_df=pd.DataFrame(chi2_results).T
chi2_df= chi2_df.sort_values(by='p_value')
chi2_df
df_final=df_cleaned[['is_smoker','region_southeast','is_female','bmi_category_obese','age','bmi','children']]
df_final























```
